## **Classeur les valeurs dans la dictionaire**

```python
classeur = {
    "positif": [],
    "negatif": []
}
classeur
```

```python
def trier(classeur, nombre):
    
    if nombre < 0:
        classeur["negatif"].append(nombre)
    else:
        classeur["positif"].append(nombre)
        
    return classeur
# trier c'est: classeur, ordernar, organisar, serparar, etc...
trier(classeur,-221)
```

#### EX1: Faça um programa que leia **nome e média** de um aluno, guardando também a **situação** em um **dicionario**. No final, mostre o conteudo da estrutura na tela.

```python
nome = input("Nome: ")
media = int(input(f'Media de {nome}: '))

aluno = {
    'Nome': nome,
    'Media': media
}

print(f'Nome é igual a {aluno["Nome"]}')
print(f'Media é igual a {aluno["Media"]}v')

if media < 5:
    print('Situação é igual a Reprovado')
else:
    print('Situação é igual a Aprovado')
```
