# Criar graficos com Matplotlib

#### O Matplotlib - é uma lib do python usada para visualização de dados. Sera muito utili daqui para frente em analise dados

```python
import matplotlib.pyplot as plt
import numpy as np
```

# plot()

```python
x = np.arange(1,5)
y = [2,3,4,3]

display(x)
plt.plot(x,y)
plt.grid()
plt.yticks(y)
plt.xticks(x)
plt.show()
```

```python
x = np.arange(10)
y = 2*x
print(f'X = {x}')
print(f'Y = {y}')

plt.title(label='Grafico da relação entre x e y')
plt.plot(x, y,label = 'Dados')
plt.ylabel('Eixo Y')
plt.xlabel('Eixo X')
plt.xticks(x)
plt.yticks(y)
plt.legend()
plt.grid()
plt.show()
```

```python
plt.plot(x,y, c = 'red', label='dados')
plt.ylabel('Eixo y')
plt.xlabel('Eixo x')
plt.yticks(y)
plt.xticks(x)
plt.legend()
plt.grid()
plt.show()
```

```python
x = np.linspace(0, 20,200)
y = x** 2
plt.plot(x, y,color = 'red', label = 'Data')
plt.title('Traçons des graphiques avec Matplotlib')
plt.ylabel('axis y')
plt.xlabel('axis x')
plt.grid()
plt.legend()
plt.show()
```

### 'go:' - c'est une parametre qu'on va utiliser port ajouter des points sur notre graphique

```python
x = np.linspace(0, 20,20)
y = x** 2
plt.plot(x, y, 'go:',color = 'red', label = 'Data')
plt.title('Traçons des graphiques avec Matplotlib')
plt.ylabel('axis y')
plt.xlabel('axis x')
plt.grid()
plt.legend()
plt.show()
```

```python
y2 = 2*x
plt.plot(x, y,color = 'green', label = 'x²')
plt.plot(x, y2,color = 'red', label = '2x')

plt.title('Graphique montrant deux functions')
plt.ylabel('axis y')
plt.xlabel('axis x')
plt.grid()
display(y2)
plt.legend()
plt.show()
```

# Propriedade Linestyle

```python
plt.plot(np.arange(0,5), np.arange(5,10), 'r', linestyle = '--', lw = 9)
plt.grid()
plt.show()
```

# Grafico de disperção ( .scatter(x,y...))

# Grafico de Barra

```python
reseaux_sociaux = ["Facebook", "YouTube", "Whatsapp" ]
perc_use = np.array([10, 40, 60])
plt.bar(reseaux_sociaux, perc_use, color = 'r')
plt.ylabel('Redes sociais')
plt.xlabel('Useres')
plt.title('Visualizando o uso das redes sociais')
plt.grid()
# plt.gr id()
plt.show()
```

```python
plt.pie(perc_use, labels = reseaux_sociaux)
plt.show()
```

```python
plt.pie(perc_use, labels = reseaux_sociaux)
plt.title('A percentagem de uso de cada rede social')
plt.show()
```

```python
x = np.linspace(0, 2, 10)
y = x**2
# display(y)
# plt.figure(figsize=(12, 8))
plt.plot(x, y, label = 'Quadrado', c = 'red')
plt.plot(x, x**3, label = 'Cubo', c = 'blue', marker = '.')
plt.grid()
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')

plt.xticks(x)
print('X = ', x)
print('\n Y = ', y)

print('\n x**3 = ', x**3)
plt.legend()
plt.show()
plt.savefig('Jeovani.png')
```

```python
plt.plot(x, y, 'go:')
plt.grid()
plt.show()
```

```python
x = np.linspace(0,20, 20)
y1 = x ** 2
y2 = x * 2

plt.plot(x, y1, 'b', label = 'function x²')
plt.plot(x, y2, 'r', label = 'function 2x')

plt.xlabel('axes x')
plt.ylabel('axes y')
plt.title("Graphique montrant deux fonctions")
plt.grid()
plt.legend()
plt.show()
```

```python
np.random.seed(0)
x = np.random.binomial(n = 20, p = 0.2, size = 1000)
plt.title("Graphique de l'histrogramme")
plt.hist(x)
plt.show()
```

```python
x = np.arange(1,5)
display(x)
y = [1,4,2,3]
figure = plt.figure(figsize = (5,3))

figure.add_subplot(131)
figure.suptitle('Titulo da figura')
plt.plot(x,y, label = 'Qualquer coisa')
plt.legend()
plt.show(figure)
```

# Subplots
Serve para por uma ou mais graficos numa so figura

```python
vls_x = np.array(np.arange(1,5))
vls_y = np.array([1,4,2,3])
display(vls_x)
display(vls_y)

figura = plt.figure(figsize=(20,5))
figura.add_subplot(131)
plt.plot(vls_x, vls_y, label = 'Um dado Qualquer')
plt.scatter(vls_x,vls_y)
figura.suptitle('Titulo Geral')
plt.legend()
plt.grid()
plt.title('Grafico1')

# criando a  figura3

figura.add_subplot(133)
plt.title('Grafico3')
plt.scatter(vls_x, vls_y, label = 'Teceiro dado quqlquer')
plt.grid()
plt.legend()

# criando o grafico2 
figura.add_subplot(132)
plt.bar(vls_x, vls_y)
plt.title('Grafico2')

# salvando esse mesmo grafico
plt.savefig('grafico.png') # salvando na mesma pasta
plt.show()
```
