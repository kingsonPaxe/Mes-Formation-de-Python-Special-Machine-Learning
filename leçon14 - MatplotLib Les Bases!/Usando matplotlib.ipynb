# criando o meu primeiro grafico

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(10)
display(x)
y = 2*x
display(y)

plt.plot(x,y,label = 'dados')
plt.grid()
plt.xlabel("Eixo X")
plt.ylabel("Eixo y")
plt.legend()
plt.yticks(y)
plt.show()
```

```python
redes = ['Facebook', 'Instagram', 'Whatsapp']
using_persent = [30, 10, 60]

figure = plt.figure(figsize=(20,5))
figure.add_subplot(131)
figure.suptitle('Todos os graficos')

# grafico1
plt.plot(x,y, label= 'Dado 1')
plt.legend()
plt.xlabel('Axis x')
plt.ylabel('Axis y')
plt.title('Grafico1')
plt.grid()

# grafico2
figure.add_subplot(132)
plt.bar(redes,using_persent)
plt.xlabel('Axis x')
plt.ylabel('Axis y')
plt.title('Grafico2')


# grafico de pizza
figure.add_subplot(133)

plt.pie(using_persent, labels = redes)
plt.title('Redes mais usadas')

plt.show()
```

```python
figure = plt.figure(figsize=(10,10) )
figure.add_subplot(2,1,1)
x = np.linspace(0, 2, 20)
y = x**2
plt.grid()
plt.plot(x,y, label = 'relação entre x & y')
plt.plot(x, x**3, label = 'x+3')
plt.xlabel('Eixo x')
plt.ylabel('Eixo Y')
plt.title('Grafico linear')
plt.legend()

figure.add_subplot(2,1,2)
plt.grid()
plt.plot(x,y, label = 'relação entre x & y')
plt.plot(x, np.sin(x), label = 'sin(x)', c = 'red')
plt.xlabel('Eixo x')
plt.ylabel('Eixo Y')
plt.title('Grafico linear')
plt.legend()
plt.show()

plt.show()
```

# Exercices

vous avez un dictionaire qui comprend **N datasets**.
Créez une fonction qui trace **N graphiques** sur une seule figure

```python
experiance = {f'Experiance{i}':np.random.randn(100) for i in range(4)} # dict comprehension
# display(experiance)
experiance
```

```python
figure = plt.figure(figsize=(10,10))
for pos in range(1,4): 
    figure.add_subplot(3,1,pos)
    plt.plot(experiance[f'Experiance{pos-1}'], label = f'Experiance {pos-1}')
    plt.legend(loc = 0)
    plt.title(f'Experiance {pos-1}')
    print(pos-1)
plt.show(figure)
```

```python
figure = plt.figure(figsize = (10,10))

# Experiance 0
figure.add_subplot(3,1,1)
plt.plot(experiance['Experiance0'], label = 'Experiance 0')
plt.legend()
plt.title('Experiance 0')


# Experiance 1
figure.add_subplot(3,1,2)
plt.plot(experiance['Experiance1'], label = 'Experiance 1')
plt.legend()
plt.title('Experiance 1')

# Experiance 3
figure.add_subplot(3,1,3)
plt.plot(experiance['Experiance2'], label = 'Experiance 2')
plt.legend()
plt.title('Experiance 2')
plt.savefig('Des Experiances.png')
plt.show()
```

```python
# Fazendo usando funções
def graphiques(dataset):
       figure = plt.figure(figsize=(10,10))
       for id in range(1,4):
              figure.add_subplot(3,1,id)
              plt.plot(dataset[f'Experiance{id - 1}'], label = f'Experiance{id - 1}')
              plt.title(f'Experiance{id - 1}')
       plt.show()

graphiques(experiance)
```
