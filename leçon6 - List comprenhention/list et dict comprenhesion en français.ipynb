# 1 - **En étudiant les liste comprehension**

## 1.1 - **En commencant seul avec les boubcle for**

```python

list1 = []
for i in range(10):
    list1.append(i**2)
print(list1)
%time
```

## 1.2 - **En utilisant les list comprehension**

```python

list2 = [result **2 for result in range(10)]
list2
%time
```

# **Dict comprihension**

```python
d = {
    '0': 'Pierre',
    '1': 'Jean',
    '2':'Julie',
    '3':'Sophie'
}
```

```python
prenoms = ['Pierre','Jean','Julie','Sophie']
```

```python
dic = {k:v for k,v in enumerate(prenoms)}
dic
```

```python
age = [23,29,22]

join = {prenoms:age for prenoms, age in zip(prenoms, age) if age > 24}
join
```

```python
[x**2 for x in range(0,50) if x%3==0]
```
