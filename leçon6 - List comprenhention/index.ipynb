## **List comprenhention**

```python
def for_loop(n):
    
    my_list = []
    
    for x in range(n):
        my_list.append(x)
    return my_list

for_loop(12)
```

```python
def list_comp(n):
    var = [p for p in range(n)]
    return var

list_comp(12)
```

```python
list = [54,22,15,48,332,1265,1,664,223,156]
r = []

for num in list:
    if num%2==0:
        r.append(num)
print(r)
```

```python
var = [ num for num in list if num%2==0]
print(var)
```

```python
l = [9, 10, 11, 40, 33]
var = [resultado for resultado in l if resultado%2==0]
var
```

### Cria uma list comprension que acha o quadrado de valores armazenado numa lista.

```python
lista = [2, 4, 5]
quadrado = [result**2 for result in lista]
print(quadrado)


valor_longo = [vl for vl in range(10)]
print(valor_longo)
```

## **Exercicio2**
- **Caso1**: Dobrar o valor do produto
- **Caso2**: Todos os produtos que custarem acima de 1.000 dolares, imposto de 50% sobre o valor

### **Caso1**

```python
lista_preços = [500, 1500, 2000, 100, 25]

dobro = [double*2 for double in lista_preços]
print(f'O preco_produto = {dobro}')
```

### **Caso2**

```python
produits = [prod*0.5 for prod in lista_preços if prod>1000]
f'Os impostos = {produits}'
```

```python
x = [-1, 0, 1, 2, 3, 4,10]
y = [result*2 -1 for result in x]
print(y)
```
