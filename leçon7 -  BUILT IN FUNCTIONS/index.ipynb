# any()

```python
list_bool = [False, False, True]
any(list_bool) # verifica se pelo menos tem um dado verdadeiro na lista, si tiver então o resultado é verdadeiro
```

# all()

```python
lst = [True, True, True]
all(lst) # verifica se todos são verdadeiros, mas basta ter um falso sera falso
```

# round()

```python
valor =  3.14159
print(round(valor,2)) # arrendonda um valor, com base ao numero de casas decimais, ou seja, quantas casas depos da virgula
```

```python
numeros = [1, 2, 3, 4, 5, 6]
result = [val for val in numeros if val%2 == 0]
result
```
