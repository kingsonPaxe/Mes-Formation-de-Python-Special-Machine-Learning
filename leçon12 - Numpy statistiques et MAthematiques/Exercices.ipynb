# Exercice
Standardisez la matrice A sur chaque colonne.
A - ~A/std(A)

Resultat: chaque colonne doit avoir moyenne = 0 et écart type = 1 Praticamente pediram para fazer normalização

Z= X−μ/ σ

 
Onde:


X é o valor original,

μ é a média da matriz,

σ é o desvio padrão da matriz.

```python
import numpy as np
np.random.seed(0)
A = np.random.randint(0, 100, [10, 5])
A
```

```python
media = np.median(A, axis=0)
display(media)

std = np.std(A, axis = 0)
display(std)

```

```python
res =  ((A - media)/std)
res
```

```python
res.mean(axis = 0)
```
