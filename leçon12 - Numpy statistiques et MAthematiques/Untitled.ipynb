# np.unique() - Para ver o numero de elementos que se repetem

```python
print("Array:\n",A)
print(f"Resultado: {np.unique(A, return_counts=True)}")
```

```python
values, counts = np.unique(A, return_counts=True)
display(values)
display(counts)
```

O objectivo desta parte é organizar os dados e ver quais são os numeros que se repetem deste o inicio até ao fim onde o numero que se encotra no fim é o que mais se repte.

```python
org_counts = counts.argsort()# para organizar os numerios que se repetem
org_counts
```

```python
values[org_counts]
```

```python
for i, j in zip(values[counts.argsort()], counts[counts.argsort()]):
    print(f'Valeur {i} apparait {j}')
```
