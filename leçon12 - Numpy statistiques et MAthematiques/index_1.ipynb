### **numpy.sum()**

```python
a = np.arange(4)
print("array: ",a)
print(a.sum())
```

```python
# Ex2: 
b = np.arange(25)
b
```

```python
# Como ela funciona.
x = 0
for vls in b:
    x+=vls
print(x)
```

## **numpy.prod()**

```python
k = np.array([536870910, 536870910, 536870910, 536870910])
np.prod(k)
```

```python
k2 = np.arange(1,9)
display(k2)
k2.prod()
```

```python
# como é que ela funciona...
x = 1
for vls in k2:
    x*=vls
print(x)
```

## **numpy.cusmum()**

```python
a = np.arange(1, 6)
resutlado = a.cumsum()
print(f"Array original: {a}")
print(f"Array aplicando o cumsum: {resutlado}")
```

## **numpy.cumprod()**

```python
b = np.arange(1, 4)
resutlado_2 = np.cumprod(b)
print(f"Array original: {b}")
print(f"Array aplicando o cumprod: {resutlado_2}")
```

```python
# Exemplo 2
c = np.arange(1,6)
resutlado_3 = np.cumprod(c)
print(f"Array original: {c}")
print(f"Array aplicado o cumprod: {resutlado_3}")
```
