```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import interp1d
import seaborn as sns
```

```python
x = np.arange(1,5) # ele vai contar de 1-4 descartando o ultimo valor
y = [2,3,4,3]
plt.scatter(x,y)
new_arr = np.arange(1,5)
f = interp1d(x,y, kind="cubic")
result = f(new_arr)
plt.plot(new_arr,result, c = "r")
plt.grid()

plt.show()
```

```python
plt.plot(x,y, "go:", color = "red")
plt.xticks(x)
plt.yticks(y)
```

```python
x = np.linspace(0,20,20)
display(x)
y1 = 2**x
y2 = x*2

plt.plot(x,y1, label = "function 2^x")
plt.plot(x,y2, label = "function x*2", color = "red")
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.show()
```

# Figure Size

```python
figure = plt.figure(figsize=(20,5))
figure.add_subplot(131)
plt.plot(x,y)
plt.title("Grafico 1")
plt.grid()


# criando o grafico 2
figure.add_subplot(132)
plt.scatter(x,y,color = "red")
plt.title("Grafico 2")
plt.plot(x,y, color = "blue")
plt.grid()

# Aplicando interpolacao
figure.add_subplot(1,3,3)
new_x = np.arange(1,5)
f = interp1d(x,y, kind="linear")
result = f(new_x)
plt.scatter(x,y, color = "red")
plt.plot(new_x, result)
plt.title("Usando interpolação")

figure.suptitle("Executando graficamente todos os dados coletados")
plt.grid()
plt.show()
```
