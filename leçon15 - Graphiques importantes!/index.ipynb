```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
print('Done')
```

```python
data_iris = pd.read_csv("Iris.csv")
x = data_iris[data_iris['species'] == 'Iris-setosa']
y = data_iris[data_iris['species'] == 'Iris-virginica']
z = data_iris[data_iris['species'] == 'Iris-versicolor']
display(x)
display(y)
display(z)
```

```python
data_setosa = data_iris[data_iris['species'] == 'Iris-setosa']
data_virginica = data_iris[data_iris['species'] == 'Iris-virginica']
# data_setosa
data_virginica
```

```python
display(data_setosa.head())
display(data_virginica.head())
```

```python
plt.scatter(data_virginica['sepal_length'],data_virginica['sepal_width'], label = 'iris_virginica')
plt.scatter(data_setosa['sepal_length'],data_virginica['sepal_width'], label = 'iris_setosa', color = 'r')
plt.scatter(z['sepal_length'],z['sepal_width'], label = 'iris_versicolor', color = 'purple')

plt.title('Flor Iris')
plt.xlabel('sepal_length')
plt.ylabel('sepal_width')
plt.legend()
plt.show()
```

```python
plt.scatter(data_virginica['petal_length'],data_virginica['petal_width'], label = 'iris_virginica')
plt.scatter(data_setosa['petal_length'],data_virginica['petal_width'], label = 'iris_setosa', color = 'r')
plt.title('Flor Iris')
plt.xlabel('petal_length')
plt.ylabel('petal_width')
plt.legend()
plt.show()
```

```python
d_spal_legth = data_setosa[['sepal_length']]
d_sepal_width = data_setosa[['sepal_width']]

norm_spal_legth = (d_spal_legth - np.min(d_spal_legth, axis = 0)) / (np.max(d_spal_legth, axis = 0) - np.min(d_spal_legth, axis = 0))
norm_sepal_width = (d_sepal_width - np.min(d_sepal_width, axis = 0)) / (np.max(d_sepal_width, axis = 0) - np.min(d_sepal_width, axis = 0))

data_setosa['sepal_length'] = norm_spal_legth
data_setosa['sepal_width'] = norm_sepal_width
data_setosa
```
