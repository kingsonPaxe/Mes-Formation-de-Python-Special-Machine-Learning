```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
print('Done')
```

```python
data_iris = pd.read_csv("Iris.csv")
x = data_iris[data_iris['species'] == 'Iris-setosa']
y = data_iris[data_iris['species'] == 'Iris-virginica']
z = data_iris[data_iris['species'] == 'Iris-versicolor']
display(x.head())
display(z.head())
display(y)
```

```python
data_setosa = data_iris[data_iris['species'] == 'Iris-setosa']
data_virginica = data_iris[data_iris['species'] == 'Iris-virginica']
# data_setosa
data_virginica.head()
```

```python
display(data_setosa.head())
display(data_virginica.head())
```

```python
plt.scatter(data_virginica['sepal_length'],data_virginica['sepal_width'], label = 'iris_virginica', cmap="viridis")
plt.scatter(data_setosa['sepal_length'],data_virginica['sepal_width'], label = 'iris_setosa', cmap="viridis")
plt.scatter(z['sepal_length'],z['sepal_width'], label = 'iris_versicolor',cmap="viridis")

plt.title('Flor Iris')
plt.xlabel('sepal_length')
plt.ylabel('sepal_width')
plt.legend()
plt.show()
```

```python
x = np.linspace(4,100)
# y = np.arange(11,21)
y_2 = np.linspace(4,100)
# x = x.reshape(10,1)
print(f"x_menor = {x}\n")
print(f"y_menor = {y_2}\n")

X, Y= np.meshgrid(x,y_2)
print(f"X = {X}\n e tem um tamanho de {X.shape}")
print(f"Y = {Y}\n e tem um tamanho de {X.shape}")

f = lambda x,y: np.sin(x) + np.cos(y)
f_2 = lambda x,y: x + y
z = f(X,Y)

print(f"Z = {z}\n")
axes = plt.axes(projection = "3d")
axes.plot_surface(X,Y,z, cmap = "Blues")
plt.show()
```
