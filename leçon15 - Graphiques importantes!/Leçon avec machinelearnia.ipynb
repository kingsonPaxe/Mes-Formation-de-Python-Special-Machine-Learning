```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Carregar o conjunto de dados
iris = load_iris()
x = iris.data  # Dados
y = iris.target  # Classes ou tipos de flores
names = list(iris.target_names)  # Nomes das classes

def graphique(datasets, targets, colunas, linhas):
    """
    Cria gráficos de dispersão para o conjunto de dados Iris.

    Parameters:
    - datasets: dados das flores.
    - targets: classes das flores.
    - colunas: número de colunas de gráficos.
    - linhas: número de linhas de gráficos.
    """
    figure = plt.figure(figsize=(20, 5))
    figure.suptitle("Gráfico da Flor Iris", fontsize=16)

    for i in range(linhas * colunas):
        if i >= len(datasets[0]):
            break

        ax = figure.add_subplot(1, colunas, i + 1)
        
        # Scatter plot com cores baseadas nas classes
        scatter = ax.scatter(datasets[:, i % len(datasets[0])], datasets[:, (i + 1) % len(datasets[0])], c=targets, cmap='viridis')
        
        # Configuração do eixo
        ax.set_xlabel(f'Feature {i % len(datasets[0])}')
        ax.set_ylabel(f'Feature {(i + 1) % len(datasets[0])}')
        ax.set_title(f'Feature {i % len(datasets[0])} vs Feature {(i + 1) % len(datasets[0])}')

    # Adicionar legenda única com nomes das classes
    handles = [plt.Line2D([0], [0], marker='o', color='w', label=name, 
                          markerfacecolor=plt.cm.viridis(i / len(names)), markersize=10) 
               for i, name in enumerate(names)]
    figure.legend(handles=handles, loc='upper right', title="Classes")

    plt.tight_layout()
    plt.show()

# Chamando a função
graphique(x, y, 3, 1)
```

# 2 - Contour Plot

Um contour plot (gráfico de contorno) é uma representação visual que mostra as linhas de níveis (ou contornos) de uma função bidimensional. No Python, você pode criar gráficos de contorno usando a biblioteca Matplotlib. Aqui está um guia simples para utilizá-la:

## Passo a Passo para Criar um Contour Plot
### 1 - Importe as bibliotecas necessárias:

    * numpy para criar os dados;
    * matplotlib.pyplot para plotar.

### 2 - Defina os dados:

    * Crie um espaço de coordenadas (usando numpy.meshgrid);
    * Defina a função cujos contornos você quer visualizar.

### 3 - Use as funções contour ou contourf:

    * contour: cria linhas de contorno;
    * contourf: preenche as áreas entre os contornos com cores.

### 4 - Adicione legendas e rótulos (opcional):

    * Use clabel para adicionar rótulos às linhas;
    * Adicione uma barra de cores com plt.colorbar().

```python
plt.contour(X,Y,Z)
plt.show()
```

```python
plt.contour(X,Y,Z, 20)
plt.show()
```

### Para facilitar o entendimento, usamos para ver o contorno dos grafico, neste caso em 3D

```python
ax = plt.axes(projection = "3d")
ax.plot_surface(X,Y,Z, cmap = "plasma")
plt.contour(X,Y,Z)
plt.show(ax)
```

```python
plt.contourf(X,Y,Z, 20, cmap = "RdGy")
plt.xlabel("largura do sepal")
plt.ylabel("Altura do sepal")
plt.colorbar()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Criar o espaço de coordenadas
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# 2. Definir a função z = f(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# 3. Exibir o gráfico com imshow
plt.figure(figsize=(8, 6))
img = plt.imshow(Z, extent=[-5, 5, -5, 5], origin='lower', cmap='viridis', interpolation='nearest')
plt.colorbar(img, label='Valor de Z')  # Adicionar barra de cores

# 4. Configurar títulos e rótulos
plt.title("Mapa de Calor de $z = \sin(\sqrt{x^2 + y^2})$")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")

# Mostrar o gráfico
plt.show()
```

```python
# Organizando os dados
idades.sort()
age = np.array(idades)
print(f"Array organizado: {age} \n") 
# f = (4/age.size) * 100 # Mutiplicando por 100 estou obtendo uma taxa percentual
# print(f"Quer dizer que {f}% das pessoas têm 17 anos")


# Obtendo a frequencia relativa
# Consite em saber a percentagem das idades que ha nnessa lista
data = []
for a,count in zip(x,y):
    f_relativa = (count/age.size) * 100
    # onde count é a frequência absoluta e aga.size é o numero de alunos que dissera a sua idade
    """
        f_relativa = (fi/n) * 100
        onde:
            fi ou fa: frequencia absoluta ou simples que indica o numero de vezes que uma variavel se repete
            n : A quantidade de pessoas na lista 

    """
    data.append(f_relativa)
    print(f"O número {a} aparece {count} vezes e {f_relativa}% das pessoas têm {a} anos de idade") # O número de vezes que um valor aparece é chamada de frenquência absoluta ou simples
```
