# Uando os dados do Kagle
Depois de testar o database da lib sklean, decidi usar o dataset do kagle para praticar e experimentar alguns comandos

# A tarfa de hoje será converter um target em valores numerico

```python
dt = pd.read_csv('IRIS.csv')
dt.head()
```

# Criando uma variavel target

```python
target = np.array(dt['species'])
unique_target = np.unique(target)
display(unique_target)
index = 0
for t in unique_target:
    target[target == t] = index
    index+=1
print(target)
display(target.size)
```

```python
print(f'Target interno do sklearn {y}')
print(f'Target que eu cirei {target}')
display(dt)
```

# Criando o grafico

```python

x1 = dt.values # Convertendo em array
plt.scatter(x1[:,0],x1[:,1], c = target)
plt.xlabel('sepal_length')
plt.ylabel('sepal_width')
plt.title('En regardant les diffents types de sepal du fleur iris')
plt.show()
```

```python
# Controlado a trasparência do pontos no grafico com o atributo alpha.
plt.scatter(x1[:,0], x1[:,1], c = target, alpha = 0.5) # 50% da trasperencia
plt.show()
```

```python
# Aumentando o tamanho dos pontos no grafico com a função S
plt.scatter(x1[:,0], x[:,1], c = target, s = 100, alpha = 0.5)
```

```python
figure = plt.figure(figsize=(15,2))
# figure.subtitle("Grafico da Flor iris")
figure.add_subplot(131)
plt.scatter(x1[:,0], x1[:,1],c = target)
plt.xlabel('largura_sepal')
plt.ylabel('Altura_sepal')
plt.title("Fig1:Original")


figure.add_subplot(132)
plt.scatter(x1[:,0], x1[:,1],c = target, alpha = 0.5)
plt.xlabel('largura_sepal')
plt.ylabel('Altura_sepal')
plt.title("Fig2:Transparência em 50%")

figure.add_subplot(133)
plt.scatter(x1[:,0], x1[:,1],c = target, alpha = 0.5, s = 100)
plt.xlabel('largura_sepal')
plt.ylabel('Altura_sepal')
plt.title("Fig3: Aumentando a espeçura do ponto")

plt.show(figure)
```

# Grafico 3D

Uma das formas de você visualizar uns dados 3D é usando a função 



```python
from mpl_toolkits.mplot3d import Axes3D
```

```python
ax = plt.axes(projection = "3d")
ax.scatter(x1[:,0], x1[:,1],x[:,2],c = target)
plt.show()
```

```python
f = lambda x, y: np.sin(x) + np.cos(x+y)
```

```python
X = np.linspace(0,5,100)
Y = np.linspace(0,5,100)
X,Y = np.meshgrid(X,Y)
Z = f(X,Y)
```

```python
Z.size
```

```python
ax = plt.axes(projection = '3d')
ax.plot_surface(X,Y,Z, cmap= "plasma")
plt.show()
```
