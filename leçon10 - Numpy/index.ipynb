# Criação de arrays...
**np.arange()**: cria um array definindo o inicio ate ao fim


```python
a = np.arange(10)
print(a)
```

**np.identity**: Cria uma matriz identidade. onde recebe o parametro de quantas linhas deve ter

**np.ones()**: Cria uma matriz de Uns(1). Onde recebe o paramentro de quntas linhas deve ter. sempre sera mxn (linha, coluna)

```python
ones = np.ones((3,4))
print(ones)
```

**np.zeros()**: Cria uma matriz de Zeros(0). Onde recebe o paramentro de quntas linhas deve ter. sempre sera mxn (linha, coluna)

```python
zeros = np.zeros((3,4))
print(zeros)
```

# Manupulation

```python
a_ones = np.ones((3,2))
a_zeros = np.zeros((3,2))

print(f"{a_ones}")
print(f"\n{a_zeros} ")
```

### Unido dados no sentido horizontal

```python
joining_H = np.hstack((a_ones, a_zeros))

display(joining_H)
print(joining_H.shape)
print("A lista tem",joining_H.size, 'elementos')
```

### Unido dados no sentido vertical

```python
joining_V = np.vstack((a_zeros,a_ones))
print(joining_V)
print(f"\nTamanho = {joining_V.shape}")
print(f"A lista tem {joining_V.size} elementos")
```

# np.concatenate()

```python
jvc = np.concatenate((a_ones, a_zeros), axis = 0) # vertical
jhc = np.concatenate((a_ones, a_zeros), axis = 1) # horizontal

print("Vertical:\n\n",jvc) # printar a lista em vertical
print("\n horizontal: \n\n", jhc) # printar a lista em horizontal
```
