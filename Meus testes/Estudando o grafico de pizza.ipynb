# Como criar?

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
redes = ["Facebook", "Whatsapp", "Instagram"]
colors = ["red","green", "yellow"]
pessoas = [20,40,40]
plt.pie(pessoas, labels = redes, autopct='%1.1f%%', colors = colors)
plt.legend()
plt.axis("equal")
plt.show()
```

```python
plt.bar(redes, pessoas, label = "Redes sociais", color='r')
plt.legend()
plt.title("As redes sociais mais usadas pelos pessoas")

plt.show()
```
