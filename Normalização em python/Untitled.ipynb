# Usando o KNN e fazendo normalização dos dados(Standard) Z-score

```python
a = np.array([1.65, 1.85, 1.88])
p = np.array([75,90,80])

# Z-score( Standardnization)
# z = x - ~x/std(x)
tableau = np.vstack((a,p)).T
tableau

# normalizão Padronizada
tableau_normalisé = (tableau - np.mean(tableau, axis = 0))/np.std(tableau, axis = 0)
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

# Dados
altura = np.array([1.65, 1.85, 1.88]) 
peso = np.array([75, 90, 80])

# Normalização (Z-score)
scaler = StandardScaler()
X = np.column_stack((altura, peso))
X_normalized = scaler.fit_transform(X)

# Definição de labels (0 - Pequena, 1 - Grande)
# Assumindo que as duas primeiras pessoas são consideradas "grandes" e a última "pequena"
y = np.array([1, 1, 0])

# Treinamento do modelo
knn = KNeighborsClassifier(n_neighbors=1)
knn.fit(X_normalized, y)

# Previsão para novos dados
# Por exemplo, altura = 1.75m e peso = 85kg
new_data = np.array([[1.85, 80]])
new_data_normalized = scaler.transform(new_data)
prediction = knn.predict(new_data_normalized)

Alt = ['Pequeno', 'Grande']
prediction
print(Alt[prediction[0]])
print(prediction)
```
