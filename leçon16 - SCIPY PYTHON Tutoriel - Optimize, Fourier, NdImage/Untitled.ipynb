```python
from scipy.interpolate import interp1d
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,10,10) # o ultimo valor indica o numero de pontos que teremos na nossa funcao. Neste caso searo 10ponts
y = x ** 2
display(x)
plt.scatter(x,y)
plt.show()
```

```python
f = interp1d(x,y, kind="linear")
new_x = np.linspace(0,10,30)  # o ultimo valor indica o numero de pontos que teremos na nossa funcao. Neste caso 30pontis
result = f(new_x)

plt.scatter(x,y, label = "x**2")
plt.scatter(new_x,result, c= "r", label = "Function Interpolate")
plt.grid()
plt.legend()
```

```python
a = np.linspace(0,5,10)
b = a**3

plt.scatter(a,b)
```

```python
function_interpolate = interp1d(a,b, kind="linear")
nw_a = np.linspace(0,5,20)
result = function_interpolate(nw_a)

plt.scatter(a,b, color = "red")
plt.plot(nw_a,result, color = "orange")
```
