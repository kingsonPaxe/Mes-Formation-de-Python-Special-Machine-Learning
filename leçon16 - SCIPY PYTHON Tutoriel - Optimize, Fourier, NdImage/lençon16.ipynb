## Interpolate

```python
from scipy.interpolate import interp1d
```

interp1d nos retorna uma funcao: f(...) = ...

Agente pode utilizar essa funcao como queremos:
    
    *x = np.linspace(0,10,30)
    * f(x)

```python
x = np.linspace(0, 10,5)
y = x**2
plt.scatter(x,y)
plt.show()
```

```python
f = interp1d(x,y,kind='linear') # kind quer dizer tipo em portugues
nwe = np.linspace(0, 10, 30)
result = f(nwe)
plt.scatter(x,y)
plt.plot(nwe, result, c = "r")
```
