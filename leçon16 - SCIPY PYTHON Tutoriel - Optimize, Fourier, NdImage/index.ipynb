# 1. Optimize
O módulo scipy.optimize fornece funções para resolver problemas de otimização e ajuste, que são essenciais em Machine Learning.

Aplicações em Machine Learning:
Ajuste de Modelos (Curve Fitting):

Ajustar parâmetros de um modelo para minimizar o erro entre os dados previstos e os observados.
Exemplo: Ajustar os coeficientes de uma regressão polinomial.
Otimização de Funções (Gradient Descent):

Minimizar funções de perda em treinamento de modelos.
Implementação eficiente para encontrar mínimos locais/globais.
Métodos como:

minimize: Resolve problemas gerais de minimização.
least_squares: Para ajuste de dados não lineares.
root: Encontra raízes de equações.

```python
from scipy.optimize import minimize

# Função de custo
def func(x):
    return x**2 + 3*x + 2

# Encontrar o mínimo
result = minimize(func, x0=0)
print("Mínimo encontrado:", result.x)
```

# 2. Fourier
O módulo numpy.fft ou scipy.fft lida com transformadas de Fourier, que decompõem sinais em frequências.

Aplicações em Machine Learning:
Processamento de Sinais:

Análise de sinais de áudio, imagens ou séries temporais.
Identificar frequências dominantes para filtrar ou destacar padrões.
Compressão de Dados:

Compactar informações, como em imagens e sons.
Pré-Processamento:

Reduzir ruídos em dados de entrada.

```python
import numpy as np
from scipy.fft import fft, ifft

# Sinal de exemplo
signal = np.sin(2 * np.pi * np.arange(0, 1, 0.01)) + np.random.normal(0, 0.1, 100)

# Transformada de Fourier
freqs = fft(signal)

# Transformada inversa para reconstrução
reconstructed = ifft(freqs)
```

# 3. NdImage
O módulo scipy.ndimage fornece ferramentas para manipulação e análise de imagens em várias dimensões.

Aplicações em Machine Learning:
Pré-processamento de Imagens:

Redimensionamento, rotação e filtros para melhorar a qualidade do dado de entrada.
Segmentação de Imagens:

Detecção de bordas, remoção de ruídos e análise de características.
Data Augmentation:

Transformar imagens para treinar redes neurais em cenários variados.

```python
from scipy.ndimage import rotate, gaussian_filter
import numpy as np

# Criar uma imagem simples
image = np.zeros((100, 100))
image[30:70, 30:70] = 1  # Quadrado branco no centro

# Aplicar transformações
rotated_image = rotate(image, angle=45)
blurred_image = gaussian_filter(image, sigma=5)
```
