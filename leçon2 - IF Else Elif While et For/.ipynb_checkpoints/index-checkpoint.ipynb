## **If, Else, Elif**

```python
# C'est ta dire sinal en français
def signe(x):
    if x > 0:
        print(x, 'Positif')
    elif x == 0:
        print(x , 'null')
    else:
        print(x, 'Negatif')
```

```python
signe(0)
```

## **While et For**

### **For**

```python
total = 0
for count in range(10,-10,-1):
    # print('Bonjour ', count)
    signe(count)
    total+=1

print('\nTotal = ', total)
```

### **While**

```python
total = 0
element = 0
while element < 10:
    print(element)
    element+=1
    total+=1

print("Total = ", total)
```

```python
element = 10
while element >= 0:
    signe(element)
    element-=1

    if element == 0:
        print('\n')   
        break

while element <= 10:
    signe(element)
    element+=1
```
