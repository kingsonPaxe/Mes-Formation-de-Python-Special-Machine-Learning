```python
import math

def fibonacci_binet(n):
    phi = (1 + math.sqrt(5)) / 2
    return round((phi**n - (1 - phi)**n) / math.sqrt(5))

# Exemplo de uso:
for i in range(10):
    print(fibonacci_binet(i))
```

```python
total = 0
def fib(n):
    if n <=0:
        return 0
    elif n == 1:
        return 1
    else:
        result = fib(n-1) + fib(n-2)
        return (result)

valor = int(input('Digite o valor: '))
for count in range(valor):
    print(fib(count))
    total+=1

print('Total = ', total)
```
